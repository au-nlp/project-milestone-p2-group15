```python
gpt5_mini = "GPT-5-mini (high)"

df_new = df[df[MODEL_NAME]==gpt5_mini]
df_new
```

## Find all problems GPT5-mini could not solve

```python
CORRECT = "correct"
df_new_false = df_new[df_new[CORRECT]==False]
example = df_new_false.head(1)[["parsed_answer", "unique_problem_label", "answer", "user_message", "gold_answer"]]
example
```

```python
import textwrap
def prettify( f, col):
  text = df[col].iloc[0]  # e
  return  textwrap.fill(text, width=80)
print(prettify(example, "answer"))
```

```python
df_new_true = df_new[df_new[CORRECT]==True]
len(df_new_true["unique_problem_label"].unique())
```

```python
def select_problem_sample():
  """Here we just try to selct problems it could not answer"""
  problems_it_could_not_anwer = df_new_false["unique_problem_label"].unique()
  df = df_new[df_new["unique_problem_label"].isin(problems_it_could_not_anwer)]
  difficulty_levels = list(df["ten_percent_quantile"].unique())
  selected_problems = []
  for lvl in difficulty_levels:
    # select a problem with this level 
    selected_problem= df[df["ten_percent_quantile"]==lvl].head(1)
    selected_problems.append(selected_problem)
  return pd.concat(selected_problems)

selected_problems = select_problem_sample()
selected_problems
```

```python
data_path = "data/selected_problems/gpt5mini_8problems.csv"
selected_problems.to_csv(data_path)
```
