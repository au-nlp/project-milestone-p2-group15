```python
problems_path = "data/all_problems_pd/all_problems.csv"
all_problems = pd.read_csv(problems_path)
mask = ~all_problems["unique_problem_label"].str.contains("kangaroo", case=False, na=False)
all_problems = all_problems[mask]
```

```python
def select_problem_sample():
  """Here we just try to selct problems it could not answer"""
  problems_it_could_not_anwer = df_new_false["unique_problem_label"].unique()
  df = df_new_false
  difficulty_levels = list(df["ten_percent_quantile"].unique())
  selected_problems = []
  all_problems["unique_problem_label"] = all_problems["unique_problem_label"].map(lambda x: x.replace("-", "_"))
  df["unique_problem_label"] = df["unique_problem_label"].map(lambda x: x.replace("-", "_"))
  for lvl in difficulty_levels:
    # select a problem with this level 
    selected_problem= df[df["ten_percent_quantile"]==lvl].head(1)
    index = list(selected_problem["unique_problem_label"])[0]
    index = index.replace("_outputs", "").replace("-", "_")
    
    problem = all_problems[all_problems["unique_problem_label"]==index].iloc[0]["problem"]

    selected_problem["problem"] = problem

    selected_problems.append(selected_problem)
    
  return pd.concat(selected_problems)

selected_problems = select_problem_sample()
selected_problems
```
