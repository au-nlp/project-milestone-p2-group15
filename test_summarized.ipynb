```python
%load_ext autoreload
%autoreload 2
from multi_agent.multi_agent import Role
from summerized_conv import summarized_loop
solver = Role(
  name="Solver",
  behavior="Solve"

)

rejector = Role(
  name="Rejector",
  behavior="Reject"
)

summarizer = Role(
  name="Summarizer",
  behavior="Summarize"
)

clean = lambda M: [x for x in M if x]


def P(i, inst:str , M):
  if "Solve" in inst:
    return f"S{i}(" + ", ".join(clean(M))+ ")"
  if  "Reject" in inst:
    return f"R{i}(" + ", ".join(clean(M))+ ")"
  if  "Summarize" in inst:
    return f"Z{i}(" + ", ".join(clean(M))  + ")"
  else:
    raise ValueError


  
summarized_loop(n=3, P=P, solver=solver, rejecter=rejector, summarizer=summarizer)
```
